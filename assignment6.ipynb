### Question 1


```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add 2 degrees C to every reading using vectorization
temp_corrected = temperature + 2
print("a. Corrected Temperatures (Celsius):", temp_corrected)

# b. Convert all temperatures from Celsius to Fahrenheit
temp_fahrenheit = (9/5) * temperature + 32
print("b. Temperatures in Fahrenheit:", temp_fahrenheit)

# c. Identify all readings greater than 32 degrees C using Boolean indexing
greater_than_32 = temperature[temperature > 32]
print("c. Readings greater than 32 degrees C:", greater_than_32)

# d. Count how many readings exceed 32 degrees C
count_exceeding = np.sum(temperature > 32)
print("d. Count of readings exceeding 32 degrees C:", count_exceeding)

# e. Explanation of vectorization and Boolean indexing efficiency
print("\ne. Explanation:\nVectorization executes operations via optimized, compiled C code under the hood, utilizing SIMD (Single Instruction, Multiple Data) to perform parallel operations on entire arrays at once. This avoids the Python interpreter overhead of loops. Boolean indexing similarly filters elements in a single optimized pass, making data processing significantly faster and more syntactically concise than manually iterating through every single value.")
```

### Question 2


```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Calculate the total steps recorded
total_steps = np.sum(steps)
print("a. Total steps recorded:", total_steps)

# b. Calculate the mean number of steps
mean_steps = np.mean(steps)
print("b. Mean number of steps:", mean_steps)

# c. Find the maximum and minimum values
max_steps = np.max(steps)
min_steps = np.min(steps)
print(f"c. Maximum steps: {max_steps}, Minimum steps: {min_steps}")

# d. Calculate the total steps for each day using axis=0
daily_total = np.sum(steps, axis=0)
print("d. Total steps for each day (axis=0):", daily_total)

# e. Calculate the total steps for each user using axis=1
user_total = np.sum(steps, axis=1)
print("e. Total steps for each user (axis=1):", user_total)

# f. Find the user/day position containing the maximum number of steps using argmax()
flat_argmax = np.argmax(steps)
position = np.unravel_index(flat_argmax, steps.shape)
print(f"f. Flat argmax index: {flat_argmax}, Grid position (User, Day): {position}")
```

### Question 4


```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

# a. Display shape and dimensions of X
print(f"a. Shape of X: {X.shape}, Dimensions of X: {X.ndim}")

# b. Calculate X.T and explanation
XT = X.T
print("b. X.T (Transpose):\n", XT)
print("Explanation: Transpose swaps rows and columns, turning features into rows and users into columns.")

# c. Calculate matrix product X.T @ X
XTX = XT @ X
print("c. X.T @ X:\n", XTX)

# d. Calculate the matrix inverse
XTX_inv = np.linalg.inv(XTX)
print("d. Inverse of (X.T @ X):\n", XTX_inv)

# e. Implement Ordinary Least Squares: beta = (X^T X)^-1 X^T y
beta = XTX_inv @ XT @ y
print("e. Calculated coefficients (beta):", beta)

# f. Explanation of coefficients
print("\nf. Explanation of coefficients:")
print(f"   - Sleep hours coefficient ({beta[0]:.2f}): Change in score per sleep hour.")
print(f"   - Activity level coefficient ({beta[1]:.2f}): Change in score per activity unit.")
print(f"   - Stress level coefficient ({beta[2]:.2f}): Change in score per stress level point.")

# g. Prediction for new user: [5, 40, 7]
new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print(f"\ng. Predicted score for user [5, 40, 7]: {prediction:.2f}")
```
