```python
# part i
roll_number = "1024170188"
L = [int(digit)*10 for digit in roll_number]
print("L = ", L)
```

```python
# part iii
L.remove(70)
print("After removing 70: ", L) # here - the element 70 will be removed as given in the parameters of remove

L.pop()
print("After popping: ", L) # here - the last element will be popped from the list L, which is 100
```

```python
# part iv
L.sort()
print("Ascending sorted list" , L)

L.sort(reverse = True)
print("Desending sorted list", L)
```

# Question 2


```python
n = int(input("Enter Roll No.: "))
L=[]
m = n
while m>0:
  a = m%10
  a*=10
  L.append(a)
  m//=10
# print(L)

scores = tuple(L[:8])
print(scores)

high = max(scores)
high_index = scores.index(high)
low = min(scores)
low_count = scores.count(low)

print("Highest score:", high)
print("Index of highest score:", high_index)
print("Lowest score:", low)
print("Number of times lowest score appears:", low_count)

rev = scores[::-1]
print("Reversed tuple as list:", list(rev))
# Tuple cannot be reversed in place because tuples are immutable

x = int(input("Enter a score: "))

if x in scores:
    print("First occurrence index:", scores.index(x))
else:
    print("Score not present")

# scores[0] = 100
# Tuple elements cannot be changed directly because tuples are immutable, unlike lists

first, second, *remaining = scores
print(first)
print(second)
print(remaining)
```

# Question 3

```python
import random

random.seed(1024170190)

L = [random.randint(100,900) for i in range(100)]

odd = [i for i in L if i%2 != 0]
even = [i for i in L if i%2 == 0]

print("Odd numbers:", odd)
print("Count of odd numbers:", len(odd))

print("Even numbers:", even)
print("Count of even numbers:", len(even))

prime = []

for i in L:
    flag = True
    if i < 2:
        flag = False
    else:
        for j in range(2, i):
            if i%j == 0:
                flag = False
                break
    if flag:
        prime.append(i)

prime2 = [i for i in L if i in prime]

print("Prime numbers:", prime2)
print("Count of prime numbers:", len(prime2))

most = max(set(L), key=L.count)
print("Most frequent number:", most)
print("Frequency:", L.count(most))
```

# Question 4

```python
n = int(input("Enter Roll No.: "))

digits = []
m = n

while m > 0:
    a = m % 10
    digits.append(a)
    m //= 10

digits.reverse()
digits = digits[:8]

A = {i*7 for i in digits}
B = {i*9 for i in digits}

print("A:", A)
print("B:", B)

print("Union:", A.union(B))

print("Intersection:", A.intersection(B))

print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
# difference() gives elements present in one set but not the other, while symmetric_difference() gives elements present in either set but not in both

print("Symmetric Difference:", A.symmetric_difference(B))

print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))

x = int(input("Enter a value to remove from A: "))
A.discard(x)
print("A after discard:", A)
# discard() does not give an error if the value is not present, unlike remove()
```
