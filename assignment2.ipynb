# Question 1


```python
# part i
roll_number = "1024170188"
L = [int(digit)*10 for digit in roll_number]
print("L = ", L)
```

```python
# part ii
L.append(100)
print("After append: ", L) # here - the element got appended at the last of the list

L.insert(1, 20)
print("After insert: ", L) # here - the element will be inserted at whatever position of the list to which it is given in the first parameter
```

```python
# part iii
L.remove(70)
print("After removing 70: ", L) # here - the element 70 will be removed as given in the parameters of remove

L.pop()
print("After popping: ", L) # here - the last element will be popped from the list L, which is 100
```

```python
# part iv
L.sort()
print("Ascending sorted list" , L)

L.sort(reverse = True)
print("Desending sorted list", L)
```

```python
# part v
first_three = L[:3:]
print(first_three)

last_three = L[-3::]
print(last_three)
```

```python
# part vi
average = sum(L)/len(L)

print([x for x in L if x > average]) # it won't work if i just write print(x for x in L if x > average) because this is a generator expression, but when i enclose it withtin [].... it will make a new list
```
