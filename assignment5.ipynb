### Question 1: NumPy 1D-array with 5 elements and basic operations

### Question 2: Reversing array and finding most frequent values

```python
arr_to_reverse = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr_to_reverse[::-1]
print("a) Reversed array:", reversed_arr)

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

def find_frequent_and_indices(array):
    vals, counts = np.unique(array, return_counts=True)
    max_count = np.max(counts)
    most_frequent_vals = vals[counts == max_count]
    results = {}
    for val in most_frequent_vals:
        indices = np.where(array == val)[0]
        results[val] = indices
    return results

print("b) i. For array x:", find_frequent_and_indices(x))
print("b) ii. For array y:", find_frequent_and_indices(y))
```

### Question 5: Array manipulations, statistical measures, reshaping and resizing

```python
ucs420_Kamal_Minocha = np.array([[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]])
print("Original 2D Array:\n", ucs420_Kamal_Minocha)

print("Mean:", np.mean(ucs420_Kamal_Minocha))
print("Median:", np.median(ucs420_Kamal_Minocha))
print("Max:", np.max(ucs420_Kamal_Minocha))
print("Min:", np.min(ucs420_Kamal_Minocha))
print("Unique elements:", np.unique(ucs420_Kamal_Minocha))

reshaped_ucs420_Kamal_Minocha = ucs420_Kamal_Minocha.reshape(4, 3)
print("Reshaped Array (4x3):\n", reshaped_ucs420_Kamal_Minocha)

resized_ucs420_Kamal_Minocha = np.resize(ucs420_Kamal_Minocha, (2, 3))
print("Resized Array (2x3):\n", resized_ucs420_Kamal_Minocha)
```
